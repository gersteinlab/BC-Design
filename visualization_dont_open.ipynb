### structural complexity (contact order)

#### recovery

```python
# scatterplot, deprecated
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import statsmodels.api as sm
import os

# Load data
contact_orders = np.load('./test_results/SBC2-sum3-minlrdiv1-bs4-lr00002-epoch20-test/CATH4.2SurfProPiFoldDense/contact_orders.npy')
recoveries = np.load('./test_results/SBC2-sum3-minlrdiv1-bs4-lr00002-epoch20-test/CATH4.2SurfProPiFoldDense/recoveries.npy')

# Create a DataFrame
data = pd.DataFrame({'Contact Orders': contact_orders, 'Recoveries': recoveries})

# LOESS fit
frac = 0.3
lowess = sm.nonparametric.lowess(data['Recoveries'], data['Contact Orders'], frac=frac)
x_fit = lowess[:, 0]
y_fit = lowess[:, 1]

# Bootstrapping to calculate confidence intervals
n_bootstraps = 1000
predictions = np.zeros((n_bootstraps, len(x_fit)))

for i in range(n_bootstraps):
    # Resample data with replacement
    bootstrap_sample = data.sample(frac=1, replace=True)
    lowess_bs = sm.nonparametric.lowess(bootstrap_sample['Recoveries'], bootstrap_sample['Contact Orders'], frac=frac)
    predictions[i, :] = lowess_bs[:, 1]

# Calculate confidence intervals
ci_lower = np.percentile(predictions, 2.5, axis=0)
ci_upper = np.percentile(predictions, 97.5, axis=0)

# Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Contact Orders', y='Recoveries', data=data, alpha=0.5)

# Plot LOESS curve
plt.plot(x_fit, y_fit, color='red', label='LOESS Fit')

# Plot confidence intervals
plt.fill_between(x_fit, ci_lower, ci_upper, color='red', alpha=0.2, label='95% CI')

# Labels and title
plt.xlabel('Contact Order')
plt.ylabel('Recovery')
# plt.title('LOESS Fit with Varying Confidence Intervals')
plt.legend()
# plt.show()
# Path to save the figure
save_path = './figures/cath42'
file_name = 'recovery_contact_order.png'

# Create the directory if it doesn't exist
if not os.path.exists(save_path):
    os.makedirs(save_path)
plt.savefig(os.path.join(save_path, file_name))
```
